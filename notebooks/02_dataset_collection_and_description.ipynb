## Dataset Collection & Description

```python
# Ensure working directory is set to project root
import os
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

# Imports for numerical operations, DataFrame handling, and directory traversal
import os
import numpy as np
import pandas as pd
```

```python
# Define local path to UTKFace image directory
IMAGE_DIR = "/Users/cipherxishant/Downloads/archive/UTKFace"

# Initialize containers for successfully parsed records and malformed filenames
parsed_records = []
malformed_filenames = []

# List all image files ending with .jpg extension
all_files = [f for f in os.listdir(IMAGE_DIR) if f.lower().endswith('.jpg')]

# Parse each filename based on expected 'age_gender_race_date.jpg' pattern
for filename in all_files:
    try:
        parts = filename.split('_')
        # Verify that there are at least 4 underscore-separated parts
        if len(parts) < 4:
            raise ValueError(f"Filename has fewer than 4 parts ({len(parts)} parts)")
            
        # Extract integer attributes from the first three parts
        age = int(parts[0])
        gender = int(parts[1])
        race = int(parts[2])
        full_path = os.path.join(IMAGE_DIR, filename)
        
        parsed_records.append({
            'image_name': filename,
            'age': age,
            'gender': gender,
            'race': race,
            'filepath': full_path
        })
    except Exception as err:
        # Collect malformed filenames for reporting rather than silently skipping or crashing
        malformed_filenames.append((filename, str(err)))

# Print parsing summary
print(f"Total .jpg files scanned: {len(all_files)}")
print(f"Successfully parsed filenames: {len(parsed_records)}")
print(f"Malformed filenames count: {len(malformed_filenames)}")

# Display malformed filenames if any were encountered
if malformed_filenames:
    print("\nMalformed Filenames Discovered:")
    for fname, error_msg in malformed_filenames:
        print(f"  - File: '{fname}' | Reason: {error_msg}")
```

```python
# Construct Pandas DataFrame from successfully parsed records
df = pd.DataFrame(parsed_records)

# Reorder columns explicitly to match expected schema
df = df[['image_name', 'age', 'gender', 'race', 'filepath']]
```

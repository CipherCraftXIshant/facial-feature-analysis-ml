##  Final Pre-Processing Dataset

```python
# Ensure working directory is set to project root
import os
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir(".. ".strip())

# Import pandas for DataFrame handling
import pandas as pd

# Load engineered dataset
engineered_csv_path = "data/processed/utkface_engineered.csv"
df = pd.read_csv(engineered_csv_path)

print(f"Successfully loaded '{engineered_csv_path}' ({len(df)} rows, {len(df.columns)} columns).")
```

```python
# Exclude non-predictive identifier columns ('image_name', 'filepath')
# Reorder columns so target variable 'age' appears first for readability
final_columns = [
    'age', 'gender_0', 'gender_1', 'race_0', 'race_1', 'race_2', 'race_3', 'race_4',
    'eye_distance', 'mouth_width', 'face_width', 'face_height',
    'face_aspect_ratio', 'eye_to_face_ratio', 'mouth_to_face_ratio'
]

final_df = df[final_columns].copy()
print(f"Created final_df without identifier columns. Shape: {final_df.shape}")
```

```python
print("=== FINAL DATAFRAME HEAD (FIRST 10 ROWS) ===")
print(final_df.head(10))

print("\n=== FINAL DATAFRAME SHAPE ===")
print(f"Shape (rows, columns): {final_df.shape}")

print("\n=== FINAL DATAFRAME INFO & MEMORY FOOTPRINT ===")
final_df.info()

print("\n=== FINAL MISSING VALUE CHECK (ISNULL().SUM()) ===")
print(final_df.isnull().sum())
```

## 1. Project Overview
This project focuses on exploring and analyzing human facial attribute data to understand the relationship between measurable facial features and chronological age. Using the publicly available **UTKFace** dataset (comprising over 20,000 face images annotated with age, gender, and ethnicity labels), we perform structured feature extraction and exploratory data analysis.

### Scope & Boundaries
- **Scope of CE1 (Current Phase):** Covers dataset acquisition and structural understanding, image preprocessing, handling missing values and categorical features, extracting geometric facial features/measurements, performing Exploratory Data Analysis (EDA), and preparing a clean tabular dataset.
- **Scope of CE2 (Future Phase):** Model selection, training, hyperparameter tuning, age prediction modeling, evaluation metrics, and model deployment will be conducted in Course Evaluation 2 (CE2).

---

## 2. Problem Statement
Facial aging is a complex biological process characterized by non-linear physical changes in facial morphology, skin texture, and key landmark distances (such as eye-to-mouth ratio, jawline curvature, and eye spacing).

**Primary Question:** *Can measurable facial characteristics serve as reliable input features for age-related statistical analysis and future predictive modeling?*

By applying classical machine learning data preparation techniques and statistical feature extraction, we structure unorganized visual data into tabular representations. This enables rigorous correlation analysis between facial geometry and age without making exaggerated predictive claims upfront.

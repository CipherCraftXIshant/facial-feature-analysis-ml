## Dataset Path Verification

```python
# Define the local root directory path for the dataset
DATASET_PATH = "/Users/cipherxishant/Downloads/archive"

# Initialize counters and data structures for file and directory discovery
total_files_count = 0
jpg_files_count = 0
folder_structure = {}
sample_jpg_files = []

# Verify that the specified local path exists before scanning
if os.path.exists(DATASET_PATH):
    # Recursively traverse directory tree starting from DATASET_PATH
    for root, dirs, files in os.walk(DATASET_PATH):
        # Compute relative folder path from the base DATASET_PATH
        rel_folder = os.path.relpath(root, DATASET_PATH)
        
        # Filter files ending with .jpg extension
        jpg_in_current_folder = [f for f in files if f.lower().endswith('.jpg')]
        
        # Record count of .jpg files in the current folder
        folder_structure[rel_folder] = len(jpg_in_current_folder)
        
        # Accumulate total file and .jpg counts
        total_files_count += len(files)
        jpg_files_count += len(jpg_in_current_folder)
        
        # Store first 10 sample relative paths for .jpg images
        for filename in jpg_in_current_folder:
            if len(sample_jpg_files) < 10:
                sample_jpg_files.append(os.path.join(rel_folder, filename))
                
    # Print overall file counts
    print(f"Total number of files found overall: {total_files_count}")
    print(f"Total .jpg image files found: {jpg_files_count}")
    
    # Print discovered folder structure
    print("\nDiscovered Folder Structure:")
    for folder_path, count in folder_structure.items():
        print(f"  - Folder '{folder_path}': {count} .jpg files")
        
    # Print first 10 sample .jpg filenames with relative paths
    print("\nFirst 10 .jpg filenames (sample with relative paths):")
    for sample_path in sample_jpg_files:
        print(f"  - {sample_path}")
else:
    print(f"Error: Specified dataset path '{DATASET_PATH}' does not exist.")
```

## Phase 7: Categorical Data Encoding

### Understanding Categorical Encoding Techniques

In machine learning, categorical features must be converted into numerical representations before being passed into statistical or predictive models. Two primary encoding strategies are commonly used:

1. **Label Encoding (Ordinal Encoding):**
   - **Concept:** Assigns a unique integer value (e.g., 0, 1, 2, 3) to each unique category.
   - **When Appropriate:** Appropriate **only** for *ordinal* categorical variables where a natural, meaningful rank or order exists between categories (e.g., Education Level: `High School = 0`, `Bachelor = 1`, `Master = 2`, `PhD = 3`).
   - **Inappropriateness for Unordered Categories:** Applying Label Encoding to nominal (unordered) categories causes algorithms to interpret synthetic mathematical relationships (e.g., assuming Category 3 is "greater than" Category 1 or that Category 2 is the average of 1 and 3).

2. **One-Hot Encoding (Nominal Dummy Encoding):**
   - **Concept:** Creates a new binary indicator column (0 or 1) for each distinct category level.
   - **When Appropriate:** Required for *nominal* categorical variables where categories have no inherent order, rank, or quantitative relationship.

--- 

### Encoding Decision for UTKFace Features

- **Race Feature (`race`):** UTKFace defines 5 ethnicity codes (`0=White`, `1=Black`, `2=Asian`, `3=Indian`, `4=Others`). These categories are strictly **nominal and unordered**. Therefore, **One-Hot Encoding** is the mathematically correct choice. Label Encoding `race` would falsely introduce artificial ordinal distances into downstream statistical models.
- **Gender Feature (`gender`):** UTKFace defines binary gender codes (`0=Male`, `1=Female`). Since `gender` is already binary (0 or 1), keeping it as a single binary integer column is technically valid. However, applying One-Hot Encoding (producing explicit `gender_0` and `gender_1` indicator columns) ensures complete architectural consistency across categorical features and satisfies CE1 categorical encoding demonstration requirements.

```python
# Import pandas for DataFrame manipulation and get_dummies encoding
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Load missing-handled metadata CSV
csv_filepath = "data/processed/utkface_missing_handled.csv"
df = pd.read_csv(csv_filepath)

print(f"Successfully loaded '{csv_filepath}' ({len(df)} rows).")
```

```python
print("=== BEFORE ENCODING ===")
print("DataFrame Head:")
df.head()
```

```python
print("=== BEFORE ENCODING DTYPES & SHAPE ===")
print(df.dtypes)
print(f"\nBefore Encoding Shape (rows, columns): {df.shape}")
```

```python
# Perform One-Hot Encoding on 'race' and 'gender' columns
# Using pd.get_dummies with prefix parameters to generate explicit binary indicator columns
df_encoded = pd.get_dummies(
    df,
    columns=['gender', 'race'],
    prefix=['gender', 'race'],
    dtype=int
)
```

```python
print("=== AFTER ENCODING ===")
print("DataFrame Head:")
df_encoded.head()
```

```python
print("=== AFTER ENCODING DTYPES & SHAPE ===")
print(df_encoded.dtypes)
print(f"\n--- COLUMN COUNT COMPARISON ---")
print(f"Original df shape: {df.shape}")
print(f"Encoded df shape:  {df_encoded.shape}")
print(f"Original column count: {df.shape[1]}")
print(f"Encoded column count:  {df_encoded.shape[1]}")
print(f"Net new columns added by One-Hot Encoding: {df_encoded.shape[1] - df.shape[1]}")
```

```python
# Export encoded DataFrame to data/processed/utkface_encoded.csv
output_encoded_path = "data/processed/utkface_encoded.csv"
df_encoded.to_csv(output_encoded_path, index=False)

print(f"Encoded dataset saved successfully to '{output_encoded_path}' ({len(df_encoded)} rows, {len(df_encoded.columns)} columns).")
```
